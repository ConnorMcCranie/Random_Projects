```python
class Chop:
    def __init__(self, left : int = 1, right : int = 1, turn : bool = True):
        self.left = left
        self.right = right
        self.turn = turn
        self.alive = not (left == 0 and right == 0)

    def __repr__(self):
        return str([self.left, self.right]) + f', turn={self.turn}'

    def attack(self, other, give : str | int, recieve : str | int) -> None:
        # make sure the inputs are valid
        options = {'l', 'r', 'L', 'R', 'left', 'right', 0, 1}
        assert give in options and recieve in options, (
            'you need to pick valid hands to attack and recieve with'
        )
        assert self.turn and not other.turn, (
            'it needs to be the attacker\'s turn and not be the reciever\'s turn'
        )
        assert self.alive and other.alive, (
            'both player\'s need to have at least one finger remaining'
        )
        
        # determine the number of fingers to be added
        left_options = {'l', 'L', 'left', 0}
        to_add = self.left if give in left_options else self.right

        # add the fingers to the correct hand
        if recieve in left_options:
            other.left = (other.left + to_add) % 5
        else:
            other.right = (other.right + to_add) % 5
        
        # update the turn order
        self.turn, other.turn = False, True

        # check if still alive
        alive = other.is_alive()
        if not alive: print('oh he dead') 

    def is_alive(self) -> bool:
        self.alive =  not (self.left == 0 and self.right == 0) 
        return self.alive
        
    
```
