```python
class Chopsticks:
    def __init__(self, state : list[int] = [1, 1, 1, 1], turn : int = 0):
        assert len(state) == 4, 'each player has 2 hands'
        for num in state:
            assert 0 <= num < 5, 'players have up to 4 fingers out on each hand'
        player1 = sorted(state[0:2])
        player2 = sorted(state[2: 4])
        self.state = player1 + player2
        self.turn = turn
        self.player1 = player1 if self.turn == 0 else player2
        self.player2 = player2 if self.turn == 0 else player1

    def __repr__(self):
        return str(self.state)
    
    def attack(self, type : int):
        '''0 = least attack least, 1 = most attack least, 2 = least attack most
         3 = most attack most '''
        assert 0 <= type < 4, 'only 4 possible attacking moves'
        bit0, bit1 = type % 2, type // 2
        assert self.state[bit0] > 0, 'attacking hand needs non-zero # of fingers'
        
```
